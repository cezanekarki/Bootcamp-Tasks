###Reading  CSV files into a Spark DataFrame  
here we have five csv files 
1) Sales
2) customers
3)products
4)employees
5)employee_sales

```
%python
sales = spark.read.format("csv") \
    .option('header', True) \
    .option('inferSchema', True) \
    .option('delimiter', '\t') \
    .load("dbfs:/FileStore/tables/sales")
display(sales)
customers = spark.read.format("csv") \
    .option('header', True) \
    .option('inferSchema', True) \
    .option('delimiter', '\t') \
    .load("dbfs:/FileStore/tables/customers")
display(customers)
products = spark.read.format("csv") \
    .option('header', True) \
    .option('inferSchema', True) \
    .option('delimiter', '\t') \
    .load("dbfs:/FileStore/tables/products")
display(products)
employee = spark.read.format("csv") \
    .option('header', True) \
    .option('inferSchema', True) \
    .option('delimiter', '\t') \
    .load("dbfs:/FileStore/tables/employee")
display(employee)
employee_sales = spark.read.format("csv") \
    .option('header', True) \
    .option('inferSchema', True) \
    .option('delimiter', '\t') \
    .load("dbfs:/FileStore/tables/employee_sales")
display(employee_sales)




```

```
%python
# Save the 'employee' DataFrame as a Delta table named 'employee'
employee.write.format("delta").saveAsTable("employee1")
# Save the 'sales' DataFrame as a Delta table named 'sales'
sales.write.format("delta").saveAsTable("sales1")
# Save the 'products' DataFrame as a Delta table named 'products'
products.write.format("delta").saveAsTable("products1")
# Save the 'customers' DataFrame as a Delta table named 'customers'
customers.write.format("delta").saveAsTable("customers1")
# Save the 'employee_sales' DataFrame as a Delta table named ''
employee_sales.write.format("delta").saveAsTable("employee_sales1")
```

####Retrieve sales transaction details along with customer information by joining the sales and customers tables on the customer_id column.


```
%sql
select s.transaction_id,s.customer_id,s.product_id,s.transaction_date,s.amount_paid,c.name as customer_name,c.email,c.phone from sales1 s join customers1 c on s.customer_id = c.customer_id 
```

#### create a new table "sales_customer" containing sales transaction details along  with customer information by joining the sales and customers tables on the customer_id column.








```
%sql
create or replace table sales_customer as (select s.transaction_id,s.customer_id,s.product_id,s.transaction_date,s.amount_paid,c.name as customer_name,c.email,c.phone from sales1 s join customers1 c on s.customer_id = c.customer_id )
```

####Retrieve sales transaction details along with products information by joining the sales and products tables on the product_id column.

```
%sql
select s.transaction_id,s.customer_id,s.product_id,s.transaction_date,s.amount_paid,p.product_name,p.category,p.price,p.stock_quantity from 
sales1 s join products1 p on s.product_id = p.product_id
```

####create a new table as sales_product containing details from both sales and product tables by joining both tables on product_id

```
%sql
create or replace table sales_product as (select s.transaction_id,s.customer_id,s.product_id,s.transaction_date,s.amount_paid,p.product_name,p.category,p.price,p.stock_quantity from 
sales1 s join products1 p on s.product_id = p.product_id)
```

#### Q- calculate total sales amount per product and per customer

```
%sql
select customer_id,product_id,sum(amount_paid) from sales_customer group by customer_id,product_id order by customer_id
```

####Q- Top-selling products

```
%sql
select
  product_id,
  sum(amount_paid) as Total_sales
from
  sales_product
group by
  product_id
order by
  Total_sales desc
```

#####Retrieve employees transaction details along with employees_sales information by joining the employee and employee_sales tables on the employee_id column.


```
%sql

select e.employee_id,e. employee_name ,e.role,e.department,es.sale_id,es.transaction_id,es.sale_date,es.amount from employee1 e join employee_sales1 es on e.employee_id = es.employee_id
```

####Create new table employee_employeesales  containing details from both employees and employee_sales tables by joining both tables on employee_id

```
%sql
create or replace table employee_employeesales as (select e.employee_id,e. employee_name ,e.role,e.department,es.sale_id,es.transaction_id,es.sale_date,es.amount from employee1 e join employee_sales1 es on e.employee_id = es.employee_id)
```

```
%sql
select * from employee_sales1
```

#### Q-  Highest performing  employees

```
%sql
select employee_id ,sum(amount) as total_sales_amount from employee_employeesales group by  employee_id order by total_sales_amount desc 
```

#### aggregetae sales  data to analyaze monthly and yearly trends

```
%sql
 
SELECT 
    YEAR(to_date(transaction_date, 'dd/MM/yy HH:mm')) AS year,
    MONTH(to_date(transaction_date, 'dd/MM/yy HH:mm')) AS month,
    SUM(amount_paid) AS total_sales_amount 
FROM 
    sales_customer

GROUP BY 
    year, month order by month;
```
